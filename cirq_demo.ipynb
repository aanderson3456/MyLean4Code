# Quantum Computing with Google Cirq (Local)

Cirq is Google's open-source framework for quantum computing. It runs **100% locally** using standard Python on your CPU.

In this notebook, we demonstrate creating a quantum circuit that prepares a **Bell State** ($|\Phi^+\rangle = \frac{|00\rangle + |11\rangle}{\sqrt{2}}$) and simulating it locally.

```python
# Option A: Ensure cirq is available in current kernel
%pip install -q cirq numpy

import cirq
import numpy as np

# 1. Create two qubits
q0, q1 = cirq.LineQubit.range(2)

# 2. Build the circuit (Bell State: Hadamard on q0, CNOT from q0 to q1)
circuit = cirq.Circuit(
    cirq.H(q0),
    cirq.CNOT(q0, q1)
)

print("--- Quantum Circuit ---")
print(circuit)

# 3. Simulate state vector locally
simulator = cirq.Simulator()
result = simulator.simulate(circuit)

print("\n--- Final State Vector (Dirac Notation) ---")
print(result.dirac_notation(decimals=3))
```

```python
# Add measurement gates and sample 1000 repetitions
circuit.append(cirq.measure(q0, q1, key='result'))

print("--- Circuit with Measurements ---")
print(circuit)

# Sample 1000 shots locally
sample_result = simulator.run(circuit, repetitions=1000)
histogram = sample_result.histogram(key='result')

print("\n--- Measurement Histogram (1000 shots) ---")
print("Bit outcomes (0=00, 3=11):", histogram)
```
